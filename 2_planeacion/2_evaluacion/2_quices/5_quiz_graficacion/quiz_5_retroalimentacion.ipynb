# Prompt para chatGPT  

Dibújame con matplotlib, la secuencia se funciones: $y = x^2$, $y = -x^2$, $y = 4 - x^2$, y finalmente $y = |4-x^2|$, en una arreglo matricial. 

```python
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image

# Functions
x = np.linspace(-3, 3, 400)
functions = [
    ("y = x^2", x**2),
    ("y = -x^2", -x**2),
    ("y = 4 - x^2", 4 - x**2),
    ("y = |4 - x^2|", np.abs(4 - x**2)),
]

# Generate individual plots and save
paths = []
for i, (title, y) in enumerate(functions):
    plt.figure()
    plt.hlines(0, -3, 3, colors='gray', linestyles='dashed')
    plt.vlines(0, -5, 5, colors='gray', linestyles='dashed')
    plt.plot(x, y)
    plt.title(title)
    plt.xlabel("x")
    plt.ylabel("y")
    path = f"/mnt/data/plot_{i}.png"
    plt.savefig(path)
    plt.close()
    paths.append(path)

# Load images and arrange in 2x2 matrix
imgs = [Image.open(p) for p in paths]
w, h = imgs[0].size

combined = Image.new("RGB", (2*w, 2*h))
combined.paste(imgs[0], (0, 0))
combined.paste(imgs[1], (w, 0))
combined.paste(imgs[2], (0, h))
combined.paste(imgs[3], (w, h))

output_path = r"C:\Users\marco\Downloads\combined_plots.png"
combined.save(output_path)

output_path
```

# Punto sobre trazado de funciones definidas a trozos  


$$
f(x) = \begin{cases}
x-3 & si x \leq -2\\ 
-x^{2} & \text{Si} \ -2 < x < 1 \\ 
-x + 4 & \text{si } x \geq 1
\end{cases}
$$

Aquí tienes un **script de Python con Matplotlib** para graficar la función definida por partes tal como la escribiste, cuidando que:

* Cada tramo se grafique sobre su dominio correcto
* No haya “saltos falsos” en los intervalos abiertos
* Se indiquen los puntos abiertos o cerrados con círculos rellenos / vacíos


# **Código listo para usar (Matplotlib)**

```python
import numpy as np
import matplotlib.pyplot as plt

# Tramos del dominio
x1 = np.linspace(-10, -2, 400)       # x ≤ -2
x2 = np.linspace(-2, 1, 400)         # -2 < x < 1
x3 = np.linspace(1, 10, 400)         # x ≥ 1

# Funciones por tramos
y1 = x1 - 3
y2 = -x2**2
y3 = -x3 + 4

# Gráfica
plt.figure(figsize=(8,5))

# Tramo 1: x ≤ -2
plt.plot(x1, y1, label=r"$x-3$ si $x \leq -2$")

# Tramo 2: -2 < x < 1
plt.plot(x2, y2, label=r"$-x^2$ si $-2 < x < 1$")

# Tramo 3: x ≥ 1
plt.plot(x3, y3, label=r"$-x+4$ si $x\geq 1$")

# Puntos abiertos y cerrados
# En x = -2:
plt.plot(-2, -2 - 3, "ko")     # punto cerrado para el primer tramo
plt.plot(-2, -(-2)**2, "wo", markeredgecolor="k")  # punto abierto del segundo tramo

# En x = 1:
plt.plot(1, -(1**2), "wo", markeredgecolor="k")    # punto abierto del segundo tramo
plt.plot(1, -1 + 4, "ko")      # punto cerrado para el tercer tramo

# Estética
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Función definida por partes")
plt.axhline(0, color="black", linewidth=0.7)
plt.axvline(0, color="black", linewidth=0.7)
plt.grid(True, linestyle="--", alpha=0.4)
plt.legend()

plt.tight_layout()
plt.show()
```

# **¿Qué incluye este código?**

✔ Tramos correctamente definidos (≤, <, ≥)
✔ Etiquetas con LaTeX
✔ Puntos **cerrados (●)** y **abiertos (○)** llevados al gráfico
✔ Grilla y ejes resaltados
✔ Función bien definida en todo su dominio
